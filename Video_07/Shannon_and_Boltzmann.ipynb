## 1. Where Part 6 left off

The setup carries over unchanged. A message of $N$ characters over an alphabet of $U$ distinct
characters $\{x_1,\ldots,x_U\}$ with distribution $\{p_1,\ldots,p_U\}$, where character $x_j$ occurs

$$n_j = p_j N \qquad \text{times}, \qquad \sum_{j=1}^{U} n_j = N$$

The number of distinct messages — the information microstates — is the multinomial coefficient

$$\Omega_{\text{info}} = \frac{N!}{(p_1 N)!\,(p_2 N)!\cdots(p_U N)!} = \frac{N!}{n_1!\,n_2!\cdots n_U!}$$

and Boltzmann's relation applies to it exactly as it applies to a gas:

$$S = k_B \log \Omega \qquad\Longrightarrow\qquad S_{\text{info}} = k_B \log \Omega_{\text{info}}$$

with $k_B = 1.38 \times 10^{-23}$ J/K.

So one bridge is already built: $\Omega_{\text{info}} \to S_{\text{info}}$. What is missing is the
bridge from $\Omega_{\text{info}}$ to $H(\mathcal{X})$ — the function Shannon defined, which so far
in this series has lived entirely on the information-theory side. Building it is the whole of
Part 7.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import lgamma, log

K_B = 1.380649e-23          # J/K
LN2 = log(2)

def log2_factorial(k):
    """log2(k!) via lgamma -- stays finite where k! overflows float64."""
    return lgamma(k + 1) / LN2

def log2_omega(N, probs):
    """Exact log2 of the multinomial coefficient for composition round(p*N)."""
    n = [int(round(p * N)) for p in probs]
    if sum(n) != N:
        n[int(np.argmax(n))] += N - sum(n)
    return log2_factorial(N) - sum(log2_factorial(k) for k in n), n

def shannon_entropy(probs, base=2):
    p = np.asarray([q for q in probs if q > 0], dtype=float)
    return float(-np.sum(p * np.log(p) / np.log(base)))

print(f"Boltzmann constant k_B = {K_B:.6e} J/K")
```

## 2. Taking the logarithm

The strategy is direct: $\Omega_{\text{info}}$ appears inside a logarithm in Boltzmann's relation,
so take the logarithm of the multinomial coefficient and see what falls out. The base is left
**general** — written $b$ — and Section 8 shows why that generality matters.

$$\log_b \Omega_{\text{info}} = \log_b \!\left( \frac{N!}{(p_1 N)!\cdots(p_U N)!} \right)$$

A quotient splits:

$$= \log_b N! \;-\; \log_b\!\left[(p_1 N)!\,(p_2 N)!\cdots(p_U N)!\right]$$

and a product of logs splits into a sum:

$$= \log_b N! \;-\; \Big[\log_b (p_1 N)! + \log_b (p_2 N)! + \cdots + \log_b (p_U N)!\Big]$$

Every term is now the logarithm of a factorial, which is exactly what Stirling's approximation
handles.

## 3. Stirling's approximation

$$\log n! \approx n \log n - n \qquad \text{(valid for large } n\text{)}$$

Applying it to every term and substituting $n_j = p_j N$:

$$\log_b \Omega_{\text{info}} \approx \big(N \log_b N - N\big)
- \sum_{j=1}^{U} \big(p_j N \log_b (p_j N) - p_j N\big)$$

The $-N$ and $+\sum_j p_j N = +N$ cancel. Splitting $\log_b(p_j N) = \log_b p_j + \log_b N$:

$$\log_b \Omega_{\text{info}} \approx N \log_b N - N\sum_j p_j \log_b p_j
- N \log_b N \underbrace{\sum_j p_j}_{=\,1}$$

The two $N \log_b N$ terms cancel, and what survives is

$$\boxed{\;\log_b \Omega_{\text{info}} = -N \sum_{j=1}^{U} p_j \log_b p_j = N \cdot H(\mathcal{X})\;}$$

> **On the base.** Stated as $\log n! \approx n\log n - n$, Stirling is exact only in natural log;
> the general-base form carries a factor, $\log_b n! \approx n\log_b n - n\log_b e$. It makes no
> difference here — the linear term cancels against $\sum_j n_j = N$ whichever convention is used,
> so the boxed result is unaffected. The cancellation is what makes the derivation base-agnostic,
> which is precisely what Section 8 exploits.

## 4. Inverting — the headline result

$\log_b \Omega_{\text{info}} = N \cdot H(\mathcal{X})$ inverts immediately:

$$\boxed{\;\Omega_{\text{info}} = b^{\,N \cdot H(\mathcal{X})}\;}$$

The number of distinct messages a source can produce is the base raised to (message length ×
entropy per character). In binary units, $b = 2$:

$$\Omega_{\text{info}} = 2^{\,N \cdot H(\mathcal{X})}$$

Read plainly: **a source of entropy $H$ bits per character, emitting $N$ characters, has about
$2^{NH}$ distinguishable messages available to it.** Entropy is the exponent that counts states.

This is also the cleanest statement yet of what Shannon entropy *is*. Part 1 derived it from
axioms, Parts 2–5 computed it, and here it turns out to be the per-character logarithm of a state
count — the same object Boltzmann was counting.

## 5. Checking it numerically — and being honest about Stirling

The relation is an *approximation*, valid for large $N$. Two things are worth separating, because
they behave very differently.

**Per character it converges quickly.** $\log_2 \Omega / N \to H(\mathcal{X})$, and by $N$ in the
thousands the agreement is a fraction of a percent.

**The count $\Omega$ itself does not.** The absolute error in $\log_2\Omega$ grows with $N$ even as
the *per-character* error shrinks — and since $\Omega$ is that logarithm exponentiated, the ratio
$\Omega_{\text{predicted}} / \Omega_{\text{true}}$ diverges. The useful form of the statement is
therefore always the per-character one.

At the small-$N$ end the approximation is simply wrong, which the $\texttt{AAB}$ example from
Part 6 shows starkly.

```python
# Part 6's AAB example -- N = 3 is far too small for Stirling
p_aab = [2 / 3, 1 / 3]
H_aab = shannon_entropy(p_aab)
omega_true_aab = 3            # 3!/(2!1!) -- enumerated in Part 6 as AAB, ABA, BAA
omega_pred_aab = 2 ** (3 * H_aab)

print("AAB  (N = 3, U = 2)")
print(f"  H(X)             = {H_aab:.4f} bits per character")
print(f"  Omega (exact)    = {omega_true_aab}")
print(f"  Omega = 2^(N*H)  = {omega_pred_aab:.3f}")
print(f"  -> off by {omega_pred_aab / omega_true_aab:.2f}x. Stirling needs large N; 3 is not large.\n")

# A realistic source, watched as N grows
probs = [0.2, 0.3, 0.5]
H = shannon_entropy(probs)
print(f"p = {probs},  H(X) = {H:.6f} bits per character\n")
print(f"{'N':>7}{'log2(Omega)':>15}{'N*H':>13}{'per-char err':>15}{'Omega ratio':>16}")
print("-" * 66)
for Nv in [10, 100, 1000, 10000, 100000]:
    lo, _ = log2_omega(Nv, probs)
    per_char_err = abs(lo / Nv - H)
    log_gap = Nv * H - lo                     # predicted minus true, in bits
    print(f"{Nv:>7}{lo:>15.2f}{Nv * H:>13.2f}{per_char_err:>15.6f}{2 ** log_gap:>16.3e}")
print("\nper-character error shrinks; the ratio of the counts themselves blows up.")
print("The meaningful statement is log2(Omega)/N -> H, not Omega = 2^(N*H) exactly.")
```

```python
Ns = np.unique(np.logspace(1, 5, 45).astype(int))
per_char = [log2_omega(int(n), probs)[0] / n for n in Ns]

fig, axes = plt.subplots(1, 2, figsize=(13, 5))

axes[0].semilogx(Ns, per_char, 'o-', color='steelblue', linewidth=2, markersize=4,
                 label='$\\log_2\\Omega / N$')
axes[0].axhline(H, color='crimson', linestyle='--', linewidth=1.5,
                label=f'$H(\\mathcal{{X}}) = {H:.4f}$ bits')
axes[0].set_xlabel('message length $N$', fontsize=11)
axes[0].set_ylabel('bits per character', fontsize=11)
axes[0].set_title('Per character: converges', fontsize=12)
axes[0].legend(fontsize=10)
axes[0].grid(alpha=0.3)

log_gap = [Nv * H - log2_omega(int(Nv), probs)[0] for Nv in Ns]
axes[1].loglog(Ns, log_gap, 'o-', color='darkorange', linewidth=2, markersize=4)
axes[1].set_xlabel('message length $N$', fontsize=11)
axes[1].set_ylabel('$N\\cdot H - \\log_2\\Omega$   (bits)', fontsize=11)
axes[1].set_title('Absolute gap in $\\log_2\\Omega$: grows\n'
                  '(so $\\Omega$ itself is off by an exponential factor)', fontsize=12)
axes[1].grid(alpha=0.3, which='both')

plt.tight_layout()
plt.show()

print(f"at N = {Ns[-1]:,}: per-character error {abs(per_char[-1] - H):.2e} bits, "
      f"but log2 gap {log_gap[-1]:.1f} bits (a factor of {2 ** log_gap[-1]:.2e} in Omega).")
```

## 6. The consistency check — and an erratum

Part 6 proved independently, via the multinomial theorem, that the **maximum** possible number of
microstates is $\Omega_{\text{tot}} = U^{\,N}$. A new relation claiming to count microstates must
reproduce that, and checking it is where the general base earns its keep.

Set $b = U$. We are free to choose any base, so choose the alphabet size:

$$\Omega_{\text{info}} = U^{\,N \cdot H(\mathcal{X})} \qquad \text{with } H \text{ now in base-}U\text{ units}$$

That last clause is the crux — $H$ must now be computed with $\log_U$, not $\log_2$. Converting:

$$\log_U p_j = \frac{\log_2 p_j}{\log_2 U}
\qquad\Longrightarrow\qquad
\boxed{\;\Omega_{\text{info}} = U^{\,\dfrac{N \cdot H(\mathcal{X})}{\log_2 U}}\;}$$

with $H$ back in bits. Now take a maximum-entropy source, where $H = \log_2 U$:

$$\Omega_{\text{info}} = U^{\,\frac{N \log_2 U}{\log_2 U}} = U^{\,N} \qquad \checkmark$$

exactly the multinomial theorem's answer. The lecture runs this for the genomic case $U = 4$, where
$H_{\max} = 2$ bits and $\log_2 4 = 2$, giving $\Omega = 4^N$.

> **The erratum.** Doing this check is how the lecturer found a mistake in his own work:
> *"I discovered a tiny error in my previously published papers and the book … when one operates in
> binary units, so $b=2$, base 2 logarithm, that error goes away … but actually if one wants to
> generalize properly this, there is a tiny error in the manuscripts."*
>
> The missing piece is the $\log_2 U$ conversion above. At $b = 2$ it is $\log_2 2 = 1$ and vanishes,
> which is why the published binary results are unaffected; away from base 2 it does not. He states
> he is correcting it on camera, in this lecture.

### Presenter's Notes — Changing base to $U$, and the corrected general form
![Board showing the base change to U with the log base 2 of U conversion factor in the boxed result](Screenshots/Part_7_00_48_30.jpg)

*Screenshot at timestamp 00:48:30*

---

```python
def omega_from_entropy(N, probs, b):
    """Omega = b^(N * H_b),  with H expressed in base-b units."""
    return b ** (N * shannon_entropy(probs, base=b))

# The maximum-entropy check, across several alphabet sizes
print(f"{'U':>3}{'N':>6}{'H_max (bits)':>15}{'U^(N H / log2 U)':>20}{'U^N':>16}{'match':>8}")
print("-" * 68)
for U in (2, 3, 4, 8):
    uniform = [1 / U] * U
    for Nv in (5, 12):
        H_bits = shannon_entropy(uniform, base=2)
        via_entropy = U ** (Nv * H_bits / np.log2(U))
        direct = float(U ** Nv)
        ok = np.isclose(via_entropy, direct, rtol=1e-9)
        print(f"{U:>3}{Nv:>6}{H_bits:>15.4f}{via_entropy:>20.1f}{direct:>16.1f}{str(ok):>8}")
        assert ok

print("\nEvery maximum-entropy case reproduces the multinomial theorem's U^N exactly.")

# and the base-invariance the general form promises
Nv, probs_g = 12, [0.1, 0.2, 0.3, 0.4]
vals = {b: omega_from_entropy(Nv, probs_g, b) for b in (2, 3, 4, 10)}
print(f"\nBase invariance for p = {probs_g}, N = {Nv}:")
for b, v in vals.items():
    print(f"  b = {b:>2}:  Omega = {v:.6f}")
assert np.allclose(list(vals.values()), list(vals.values())[0], rtol=1e-9)
print("  all bases agree -- b^(N*H_b) is independent of b, as it must be.")
```

## 7. The link to physical entropy

Feeding the result into Boltzmann's relation gives the lecture's closing statement. Writing the
outer logarithm in an arbitrary base $a$:

$$S_{\text{info}} = k_B \log_a \Omega_{\text{info}}
= k_B \log_a\!\left(b^{\,N H(\mathcal{X})}\right)$$

$$\boxed{\;S_{\text{info}} = k_B \cdot N \cdot H(\mathcal{X}) \cdot \log_a b\;}$$

The $\log_a b$ is pure unit bookkeeping between the two conventions:

| Choice | Factor | Result |
|---|---|---|
| $a = b$ (same base throughout) | $\log_b b = 1$ | $S = k_B N H$ |
| $a = e$, $b = 2$ (natural log, bits) | $\ln 2 \approx 0.693$ | $S = k_B N H \ln 2$ |

The second row is the physically standard one: Boltzmann entropy in J/K, Shannon entropy in bits.
So the thermodynamic entropy of an $N$-character message with $H$ bits per character is
$k_B N H \ln 2$ — a number set entirely by the message's information content and the choice of
units.

### Presenter's Notes — The final result — $\Omega_{\text{info}} = b^{NH}$ and $S_{\text{info}} = k_B N H \log_a b$
![Board showing the two boxed final relations linking microstates, Shannon entropy and Boltzmann entropy](Screenshots/Part_7_00_49_20.jpg)

*Screenshot at timestamp 00:49:20*

---

```python
def S_info(N, H_bits, a=np.e, b=2):
    """Boltzmann entropy of a message, in J/K. Default: natural log outer, bits inner."""
    return K_B * N * H_bits * (np.log(b) / np.log(a))

# the two conventions from the table
N_demo, H_demo = 8192, 1.0
same_base = K_B * N_demo * H_demo                       # a = b
nat_bits = S_info(N_demo, H_demo)                       # a = e, b = 2

print(f"a message of N = {N_demo:,} characters at H = {H_demo} bit/character\n")
print(f"  a = b        : S = k_B N H          = {same_base:.4e}  (k_B units)")
print(f"  a = e, b = 2 : S = k_B N H ln2      = {nat_bits:.4e} J/K")
print(f"  ratio        = {nat_bits / same_base:.6f} = ln 2 = {LN2:.6f}\n")

# a few concrete messages
print(f"{'message':<34}{'N':>16}{'H (bits/ch)':>13}{'S_info (J/K)':>16}")
print("-" * 79)
for label, Nv, Hv in [
    ("1 KB, maximally random bits", 8 * 1024, 1.0),
    ("1 MB, maximally random bits", 8 * 1024 ** 2, 1.0),
    ("human genome, 3.1e9 bases", int(3.1e9), 2.0),
    ("human genome at measured H~1.9", int(3.1e9), 1.9),
]:
    print(f"{label:<34}{Nv:>16,}{Hv:>13.2f}{S_info(Nv, Hv):>16.3e}")

print("\nThese are minute by thermodynamic standards -- k_B is 1.38e-23 J/K, so even a")
print("genome's worth of information carries an entropy around 1e-13 J/K.")
```

## 8. What the relation does and does not settle

The lecture closes by referencing a public criticism from physicist Sabine Hossenfelder, who argued
that Shannon information entropy *"is nothing else but Boltzmann entropy"*. The lecturer disagrees —
*"the relationship between them is not quite the same thing … they're fundamentally very
different"* — and says he will devote a fuller lecture to the distinction.

This notebook does not adjudicate that. What can be said precisely is what the mathematics above
establishes and what it leaves open:

**What is established.** For this specific construction — messages built from $N$ characters over a
$U$-letter alphabet, with microstates defined as distinct arrangements — the two quantities are
proportional, $S_{\text{info}} = k_B N H \log_a b$. Given the identification of "message
arrangement" with "microstate" that Part 6 postulated, this follows.

**What is not.** Proportionality within a construction is not identity of concepts. The open
questions are the interpretive ones: whether a message's arrangements are physical microstates in
the same sense a gas's are; whether the postulate of Part 6 is a definition or a physical claim; and
what carries the thermodynamic consequences — the substrate storing the message, or the abstract
message itself.

Those are exactly the questions the disagreement is about, and they are not resolved by the
algebra.

> **Answered in Part 8.** [Part 8: Shannon vs Boltzmann Entropy](../Video_08/Shannon_vs_Boltzmann.ipynb) is the promised lecture. Its answer to the first
> question is that Boltzmann's microstate count is the *same multinomial coefficient* used here
> for messages, but that every symbol in it refers to something different — particles vs
> letters, phase-space cells vs alphabet symbols — and that $S$ is extensive with units of J/K
> while $H$ is intensive and dimensionless. The verdict there: connected by the relation derived
> above, but not the same quantity. The deeper question of whether shared structure licenses
> treating information as physical is left to the lectures on mass-energy-information
> equivalence.

## Summary

| Relation | Meaning |
|---|---|
| $\Omega_{\text{info}} = \dfrac{N!}{n_1!\cdots n_U!}$ | microstates of a message (Part 6) |
| $\log_b \Omega_{\text{info}} = N \cdot H(\mathcal{X})$ | Stirling applied to that coefficient |
| $\Omega_{\text{info}} = b^{\,N \cdot H(\mathcal{X})}$ | **the headline** — entropy is the exponent counting states |
| $\Omega_{\text{info}} = U^{\,N H / \log_2 U}$ | the same in base $U$, with $H$ in bits (the erratum) |
| $\Omega_{\text{info}} = U^{\,N}$ at $H = \log_2 U$ | recovers Part 6's multinomial theorem |
| $S_{\text{info}} = k_B N H(\mathcal{X}) \log_a b$ | **the link** to Boltzmann entropy |

### Key takeaways

1. **Take the log of the multinomial coefficient, apply Stirling, and Shannon's $H$ appears.** The
   $N\log N$ and linear terms cancel, leaving $-N\sum_j p_j \log_b p_j$ exactly.
2. **$\Omega_{\text{info}} = b^{NH}$.** A source of entropy $H$ emitting $N$ characters has $b^{NH}$
   distinguishable messages — entropy is the exponent that counts states.
3. **It is asymptotic.** Per character it converges fast; the count $\Omega$ itself is off by an
   exponentially large factor even at large $N$. Quote the per-character form.
4. **Setting $b = U$ recovers $U^N$ at maximum entropy**, matching Part 6's independent result — the
   check that validates the whole derivation.
5. **That check surfaced an erratum** in the lecturer's published papers and book: the $\log_2 U$
   base-conversion factor. It equals 1 in binary, so published binary results stand.
6. **$S_{\text{info}} = k_B N H \log_a b$** connects the two entropies, with $\log_a b$ as unit
   bookkeeping — $\ln 2$ for J/K against bits.
