## 1. First — a correction to Part 4

Dr. Vopson begins this lecture by revisiting the closing claim of Part 4. In his words, it was
*"not a mathematical error, not a conceptual error — more like an interpretation."*

Part 4 ended by observing that building a set of $m$-blocks **increased** the total information of
the message, and read that as an argument for single characters being the better choice for
compression. The clarification:

> That is a true statement **only if the step size** used to generate the $m$-blocks **is smaller
> than the $m$-block itself** … because you are overlapping the $m$-blocks and counting many
> characters multiple times because of this overlap.

That is precisely the point. When $SS < m$ the blocks overlap, each character is read into more
than one block, and the total is inflated by the duplication rather than by anything the blocking
revealed. When $SS = m$ the blocks tile the message, every character is used exactly once, and the
inflation disappears.

> **Note.** [Part 4, Section 8](../Video_04/Information_Entropy_of_m_blocks_Examples.ipynb) of this
> series worked through exactly this reconciliation — showing that with $SS = m$ the total *falls*
> from 8.81 bits to 7.61, and that per character $h_2 = 0.765 < 0.881$. That section was written
> before this lecture was available; it is left as it stands, and the lecturer's own clarification
> here confirms the reading. Credit for the correction is his.

Everything below uses $SS = 1$, so the overlap effect is live throughout — and Section 8 returns
to it with the numbers.

## 2. The genomic alphabet

A DNA or RNA sequence — double helix or single — reduces, for information-theoretic purposes, to a
string over four distinct characters:

$$\mathcal{X} = \{A,\; C,\; G,\; T\} \implies U = 4$$

with a probability distribution $P = \{p_A,\, p_C,\, p_G,\, p_T\}$. A real genome has an enormous
number of characters while still having only these four distinct ones, so $N \gg U$.

The ceiling follows immediately:

$$H(\mathcal{X})_{\max} = \log_2 U = \log_2 4 = 2 \text{ bits per nucleotide}$$

Each nucleotide encodes **at most 2 bits**, which is what makes the two-bit binary encoding natural:

| Nucleotide | Bits |
|---|---|
| $A$ | `00` |
| $C$ | `01` |
| $G$ | `10` |
| $T$ | `11` |

> **A note on the encoding.** This lecture assigns $G = \texttt{10}$ and $T = \texttt{11}$;
> [Part 2](../Video_02/Information_Entropy_Examples.ipynb) used the opposite pairing
> ($T = \texttt{10}$, $G = \texttt{11}$). The choice is arbitrary — any bijection between four
> letters and the four 2-bit patterns works, and no entropy calculation depends on which one you
> pick. The lecture's own assignment is used here.

The lecture makes a practical point about this encoding: you *can* translate the sequence to bits
and analyse the resulting binary string as in Part 4, but you do not need to. Working directly on
the four-letter alphabet gives the same answer with less bookkeeping.

### Presenter's Notes — The nucleotide alphabet, $U = 4$, and the 2-bit encoding
![Board showing the ACGT alphabet, U equals 4, maximum entropy 2 bits, and the two-bit encoding](Screenshots/Part_5_00_07_05.jpg)

*Screenshot at timestamp 00:07:05*

---

```python
import numpy as np
import matplotlib.pyplot as plt
from collections import Counter

def shannon_entropy(probabilities, base=2):
    """Shannon entropy H = -sum(p * log_b(p)).  Skips zero-probability terms."""
    p = np.asarray(probabilities, dtype=float)
    p = p[p > 0]
    assert np.isclose(p.sum(), 1.0), f"Probabilities must sum to 1 (got {p.sum()})"
    return -np.sum(p * np.log(p) / np.log(base))

def entropy_of_message(message, base=2):
    """Compute H(X) directly from a sequence of symbols (string or list)."""
    N = len(message)
    counts = Counter(message)
    probs = [c / N for c in counts.values()]
    return shannon_entropy(probs, base=base), counts

ALPHABET = ['A', 'C', 'G', 'T']
U = len(ALPHABET)
ENCODING = {'A': '00', 'C': '01', 'G': '10', 'T': '11'}

H_max_single = np.log2(U)
print(f"alphabet X = {{{', '.join(ALPHABET)}}},  U = {U}")
print(f"H(X)_max = log2({U}) = {H_max_single:.0f} bits per nucleotide\n")
for nt in ALPHABET:
    print(f"  {nt} = {ENCODING[nt]}")
```

## 3. A hypothetical 18-nucleotide segment

The lecture uses a short randomly-chosen segment — explicitly *not* a real genome, just enough
characters to work numerically by hand:

$$\text{sequence} = \texttt{CACTTATCATTCTGACTG} \qquad N = 18$$

Counting each nucleotide gives the distribution

$$P = \left\{ p_A = \tfrac{4}{18},\; p_C = \tfrac{5}{18},\; p_G = \tfrac{2}{18},\; p_T = \tfrac{7}{18} \right\}$$

The mental model for these probabilities, as the lecture puts it: scatter all eighteen letters face
down on a table, close your eyes, and draw one. The chance of drawing an $A$ is $4/18$.

### Presenter's Notes — The 18-nucleotide segment
![Board showing the eighteen character genomic sequence CACTTATCATTCTGACTG](Screenshots/Part_5_00_08_40.jpg)

*Screenshot at timestamp 00:08:40*

---

```python
sequence = "CACTTATCATTCTGACTG"
N = len(sequence)

counts_single = Counter(sequence)
print(f"sequence = {sequence}")
print(f"N = {N} nucleotides\n")
print(f"{'nucleotide':>11}{'count':>7}{'probability':>16}")
print("-" * 34)
for nt in ALPHABET:
    c = counts_single[nt]
    print(f"{nt:>11}{c:>7}{f'{c}/{N} = {c / N:.4f}':>16}")
assert sum(counts_single.values()) == N
print(f"\ntotal {sum(counts_single.values())} -- matches N")

# the binary translation the lecture mentions but does not need
bits = ''.join(ENCODING[nt] for nt in sequence)
print(f"\nbinary translation ({len(bits)} bits = 2 x {N}):")
print(f"  {bits}")
```

## 4. Entropy of the single-character set

Applying the Shannon formula over the $U = 4$ distinct nucleotides:

$$H(\mathcal{X}) = -\sum_{j=1}^{4} p_j \log_2 p_j = 1.878 \text{ bits per nucleotide}$$

against the ceiling of 2 bits. The total information in the segment is then

$$\text{Inf} = N \cdot H(\mathcal{X}) = 18 \times 1.878 = 33.8 \text{ bits}$$

$$\text{Inf}_{\max} = N \cdot H(\mathcal{X})_{\max} = 18 \times 2 = 36 \text{ bits}$$

So the segment carries 33.8 of a possible 36 bits — close to the maximum, because the four
nucleotides appear at fairly similar rates. This is the **single-character** baseline; the rest of
the lecture asks what changes when we read the sequence in codons instead.

```python
H_single, _ = entropy_of_message(sequence)
Inf_single = N * H_single
Inf_single_max = N * H_max_single

print(f"H(X)        = {H_single:.4f} bits per nucleotide   (lecture: 1.878)")
print(f"H(X)_max    = {H_max_single:.4f} bits per nucleotide")
print(f"efficiency  = {H_single / H_max_single:.1%} of the maximum\n")
print(f"Inf         = N x H(X)     = {N} x {H_single:.4f} = {Inf_single:.2f} bits   (lecture: 33.8)")
print(f"Inf_max     = N x H(X)_max = {N} x {H_max_single:.0f} = {Inf_single_max:.0f} bits")
```

## 5. Why block at all? Codons

The lecture pauses here on a question worth asking directly: why group characters at all?

The answer it gives is that single-character counting **ignores correlations between neighbouring
characters**. Reading a genome one nucleotide at a time measures how often each letter appears, but
says nothing about which letters follow which — and in a genomic sequence that ordering is where
the biology lives.

For genomes there is a natural block size. A **codon** is a sequence of three consecutive
nucleotides; it is the unit the cell actually reads, specifying one amino acid or signalling the
end of protein synthesis during translation. So take

$$m = 3$$

and Part 3's counting relationship gives the size of the codon alphabet:

$$\Omega = U^{\,m} = 4^3 = 64$$

sixty-four distinct codons — which is exactly the size of the real genetic code. Of those 64, **61
code for amino acids and 3 are stop codons**.

The maximum entropy follows from Part 3's scaling law:

$$H(\mathcal{X})^{(m)}_{\max} = \log_2 \Omega = \log_2 U^{\,m} = m \log_2 U = 3 \times 2 = 6
\text{ bits per codon}$$

Each codon can carry at most 6 bits — three nucleotides at 2 bits each, exactly as the linear
scaling requires.

### Presenter's Notes — $m = 3$ codons: $\Omega = 4^3 = 64$ and $H_{\max} = 6$ bits
![Board showing m equals 3 codons, omega equals 64, and maximum entropy of 6 bits](Screenshots/Part_5_00_17_45.jpg)

*Screenshot at timestamp 00:17:45*

---

```python
m = 3
Omega = U ** m
H_max_codon = np.log2(Omega)

print(f"m       = {m}  (codons)")
print(f"Omega   = U^m = {U}^{m} = {Omega} distinct codons")
print(f"H_max   = log2(Omega) = {m} x log2({U}) = {H_max_codon:.0f} bits per codon\n")
print(f"of the {Omega} codons in the real genetic code:")
print(f"  61 code for amino acids")
print(f"   3 are stop codons")
```

## 6. Generating the codon set

With $N = 18$, $m = 3$ and a step size of $SS = 1$, Part 4's counting formula predicts

$$N_m = 1 + \frac{N - m}{SS} = 1 + \frac{18 - 3}{1} = 16$$

The lecture chooses $SS = 1$ deliberately — not the simplified $SS = m$ case — because overlapping
blocks are where correlations between neighbouring nucleotides can show up.

Sliding a three-wide window one character at a time along $\texttt{CACTTATCATTCTGACTG}$:

$$\{\texttt{CAC},\, \texttt{ACT},\, \texttt{CTT},\, \texttt{TTA},\, \texttt{TAT},\, \texttt{ATC},\,
\texttt{TCA},\, \texttt{CAT},\, \texttt{ATT},\, \texttt{TTC},\, \texttt{TCT},\, \texttt{CTG},\,
\texttt{TGA},\, \texttt{GAC},\, \texttt{ACT},\, \texttt{CTG}\}$$

Sixteen codons, exactly as predicted. Counting occurrences, **only two codons appear twice** —
$\texttt{ACT}$ and $\texttt{CTG}$ — and the other twelve appear once each:

$$P = \left\{ \tfrac{2}{16},\; \tfrac{2}{16},\; \underbrace{\tfrac{1}{16},\; \ldots,\; \tfrac{1}{16}}_{12 \text{ terms}} \right\}
\qquad \sum_j p_j = \frac{2 + 2 + 12}{16} = 1$$

The lecture is candid that doing this counting by hand is error-prone — it notes a slip mid-lecture
where the fractions were written over 18 (the count of *nucleotides*) instead of 16 (the count of
*codons*), and recommends software for the bookkeeping. That is precisely what the cell below does.

### Presenter's Notes — $N_m = 16$ and the full codon set
![Board showing N_m equals 16 and the list of sixteen codons generated with step size 1](Screenshots/Part_5_00_24_00.jpg)

*Screenshot at timestamp 00:24:00*

---

```python
def make_blocks(seq, m, SS=None):
    """Slide a width-m window along `seq`, advancing by SS each step (Part 4)."""
    SS = m if SS is None else SS
    if not 1 <= SS <= m:
        raise ValueError(f"step size must satisfy 1 <= SS <= m (got SS={SS}, m={m})")
    return [seq[i:i + m] for i in range(0, len(seq) - m + 1, SS)]

SS = 1
codons = make_blocks(sequence, m, SS)
N_m_predicted = 1 + (N - m) / SS

print(f"N_m predicted by formula : 1 + ({N} - {m})/{SS} = {N_m_predicted:.0f}")
print(f"N_m actually generated   : {len(codons)}")
assert len(codons) == N_m_predicted
print()
for i, c in enumerate(codons, 1):
    end = "\n" if i % 8 == 0 else "  "
    print(f"{i:>2}.{c}", end=end)

counts_codon = Counter(codons)
repeated = {c: n for c, n in counts_codon.items() if n > 1}
print(f"\ndistinct codons observed : {len(counts_codon)} of {Omega} possible")
print(f"codons appearing twice   : {', '.join(sorted(repeated))}")
print(f"probabilities sum        : {sum(n / len(codons) for n in counts_codon.values()):.4f}")
```

## 7. Entropy of the codon set

The formula is unchanged; only the alphabet it sums over has grown from $U = 4$ to $\Omega = 64$:

$$H(\mathcal{X})^{(m)} = -\sum_{j=1}^{\Omega} p_j^{(m)} \log_2 p_j^{(m)}$$

With twelve codons at $1/16$ and two at $2/16$, the sum has a pleasantly exact value:

$$H(\mathcal{X})^{(m)} = -\left[12 \cdot \tfrac{1}{16}\log_2\tfrac{1}{16}
+ 2 \cdot \tfrac{2}{16}\log_2\tfrac{2}{16}\right]
= 12 \cdot \tfrac{4}{16} + 2 \cdot \tfrac{3}{16}
= 3 + 0.75 = 3.75 \text{ bits per codon}$$

and the total information in the codon set is

$$\text{Inf} = N_m \cdot H(\mathcal{X})^{(m)} = 16 \times 3.75 = 60 \text{ bits}$$

exactly. Against a ceiling of $16 \times 6 = 96$ bits.

### Presenter's Notes — The completed board — $H(\mathcal{X})^{(m)} = 3.75$ bits and $\text{Inf} = 60$ bits
![Board showing the full calculation with codon entropy of 3.75 bits and total information of 60 bits](Screenshots/Part_5_00_32_40.jpg)

*Screenshot at timestamp 00:32:40*

---

```python
H_codon, _ = entropy_of_message(codons)
Inf_codon = len(codons) * H_codon
Inf_codon_max = len(codons) * H_max_codon

print(f"H(X)^(m)     = {H_codon:.4f} bits per codon   (lecture: 3.75)")
print(f"H(X)^(m)_max = {H_max_codon:.0f} bits per codon")
print(f"efficiency   = {H_codon / H_max_codon:.1%} of the maximum\n")
print(f"Inf          = N_m x H^(m) = {len(codons)} x {H_codon:.2f} = {Inf_codon:.0f} bits   (lecture: 60)")
print(f"Inf_max      = N_m x 6     = {len(codons)} x {H_max_codon:.0f} = {Inf_codon_max:.0f} bits\n")

# the exact hand calculation from the board
by_hand = -(12 * (1 / 16) * np.log2(1 / 16) + 2 * (2 / 16) * np.log2(2 / 16))
print(f"by hand: -[12*(1/16)log2(1/16) + 2*(2/16)log2(2/16)] = {by_hand:.4f}")
assert np.isclose(by_hand, H_codon), "hand calculation disagrees"
print("matches the direct computation exactly.")
```

## 8. Comparing the two readings of the same segment

| | single nucleotides | codons ($m=3$, $SS=1$) |
|---|---|---|
| set size | $N = 18$ | $N_m = 16$ |
| alphabet | $U = 4$ | $\Omega = 64$ |
| entropy | $1.878$ bits per nucleotide | $3.75$ bits per codon |
| maximum | $2$ bits | $6$ bits |
| total information | $33.8$ bits | $60$ bits |
| maximum possible | $36$ bits | $96$ bits |

The lecture states the headline directly, and this time with the Section 1 caveat attached: the
total rose from 33.8 to 60 bits **because $SS = 1$ counts characters more than once**. Sixteen
codons span $16 \times 3 = 48$ nucleotide-slots across a sequence of only 18 nucleotides, so each
interior nucleotide is read into roughly three different codons.

### Normalising per character

To compare across block sizes we use Part 3's per-character block entropy:

$$h_m = \frac{H(\mathcal{X})^{(m)}}{m} = \frac{3.75}{3} = 1.25 \text{ bits per nucleotide}$$

against $h_1 = 1.878$. Once the units match, the codon reading is *lower* — the direction Part 3
predicts, and the same conclusion Part 4 reached on binary data.

```python
h_per_nt_codon = H_codon / m
slots_codon = len(codons) * m

print("TOTAL INFORMATION")
print("-" * 62)
print(f"  single nucleotides   {N} x {H_single:.3f} = {Inf_single:>6.2f} bits   "
      f"({N} nucleotide-slots)")
print(f"  codons, SS = 1       {len(codons)} x {H_codon:.3f} = {Inf_codon:>6.2f} bits   "
      f"({slots_codon} nucleotide-slots -- each nucleotide ~{slots_codon / N:.1f}x)")
print()
print("PER-NUCLEOTIDE ENTROPY (like-for-like)")
print("-" * 62)
print(f"  h_1 = {H_single:.4f} bits/nucleotide   (single characters)")
print(f"  h_3 = {H_codon:.4f}/3 = {h_per_nt_codon:.4f} bits/nucleotide   (codons)")
print(f"\n  blocking lowers per-nucleotide entropy by {H_single - h_per_nt_codon:.4f} bits "
      f"({(H_single - h_per_nt_codon) / H_single:.1%})")

# the tiling case for completeness
codons_tiled = make_blocks(sequence, m, SS=m)
H_tiled, _ = entropy_of_message(codons_tiled)
print(f"\nFor completeness -- the non-overlapping reading (SS = m = 3):")
print(f"  {len(codons_tiled)} codons: {', '.join(codons_tiled)}")
print(f"  H = {H_tiled:.4f} bits/codon, total = {len(codons_tiled) * H_tiled:.2f} bits "
      f"({len(codons_tiled) * m} nucleotide-slots -- each nucleotide once)")
```

## 9. A caution these numbers demand: $\Omega \gg N_m$

There is a limit on how much any of the codon numbers can be read as a property of the *sequence*
rather than of the *sample size*, and here it bites hard.

The codon alphabet has $\Omega = 64$ states. We generated $N_m = 16$ codons. **A set of 16 samples
cannot express more than $\log_2 16 = 4$ bits of entropy**, no matter what the underlying source
does — if all 16 were distinct you would measure exactly 4 bits and no more.

We measured 3.75. That is **94% of the sampling ceiling** but only **62.5% of the true ceiling** of
6 bits. Almost the entire gap between 3.75 and 6 is the sample being too small to populate 64
states, not a discovery about the sequence.

This is the finite-sample bias quantified in
[Part 3, Section 11](../Video_03/Information_Entropy_of_m_blocks.ipynb), and the rule of thumb
given there was $\Omega \ll N$ by a factor of 10–100. Here $\Omega = 64$ against $N_m = 16$ — the
inequality points the wrong way by a wide margin.

None of this is a criticism of the lecture, which is explicit that the segment is a hand-sized
example chosen so the arithmetic can be followed. It matters for anyone tempted to run the same
procedure on real data: **estimating codon entropy needs thousands of codons, not sixteen.**

```python
sampling_ceiling = np.log2(len(codons))

print(f"Omega (possible codons)      : {Omega}")
print(f"N_m   (codons generated)     : {len(codons)}")
print(f"distinct codons observed     : {len(counts_codon)}")
print()
print(f"true ceiling      log2(Omega) = {H_max_codon:.2f} bits")
print(f"sampling ceiling  log2(N_m)   = {sampling_ceiling:.2f} bits   <- cannot exceed this")
print(f"measured                      = {H_codon:.2f} bits")
print()
print(f"  {H_codon / sampling_ceiling:.1%} of the sampling ceiling")
print(f"  {H_codon / H_max_codon:.1%} of the true ceiling")
print(f"\nPart 3's guidance: keep Omega << N.  Here Omega/N_m = {Omega / len(codons):.1f} -- "
      f"the wrong way round.")
```

```python
# How many codons would you need? Sample an unbiased 4-letter source and watch
# the estimate climb toward the true 6 bits as the sample grows.
rng = np.random.default_rng(seed=11)
sizes = np.unique(np.logspace(1, 5, 30).astype(int))
measured = []
for n in sizes:
    seq_n = ''.join(rng.choice(ALPHABET, size=n * m))
    blocks_n = make_blocks(seq_n, m, SS=m)
    h, _ = entropy_of_message(blocks_n)
    measured.append(h)

fig, ax = plt.subplots(figsize=(9, 5))
ax.semilogx(sizes, measured, 'o-', color='steelblue', linewidth=2, markersize=4,
            label='measured $H^{(m)}$')
ax.semilogx(sizes, np.minimum(np.log2(sizes), H_max_codon), 'k--', linewidth=1.5,
            label='sampling ceiling $\\log_2 N_m$')
ax.axhline(H_max_codon, color='crimson', linestyle=':', linewidth=1.5,
           label=f'true ceiling = {H_max_codon:.0f} bits')
ax.axvline(len(codons), color='darkorange', linewidth=1.5,
           label=f"this lecture's $N_m$ = {len(codons)}")
ax.set_xlabel('number of codons sampled, $N_m$', fontsize=11)
ax.set_ylabel('entropy (bits per codon)', fontsize=11)
ax.set_title('An unbiased 4-letter source, $m=3$\n'
             'the estimate tracks the sampling ceiling until the sample is large enough',
             fontsize=12)
ax.legend(fontsize=10, loc='lower right')
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

need = sizes[np.argmax(np.array(measured) > 0.99 * H_max_codon)]
print(f"For a maximally random source it takes roughly {need:,} codons "
      f"to measure within 1% of the true {H_max_codon:.0f} bits.")
```

### Key takeaways

1. **The correction to Part 4 stands at the front of this lecture**: blocking increases total
   information only when $SS < m$, because overlapping blocks read characters more than once.
2. **Genomes are a four-letter alphabet**, $U = 4$, so $H_{\max} = 2$ bits per nucleotide and the
   two-bit encoding is natural. You need not convert to binary to do the analysis.
3. **Codons are the natural $m$-block for genomes** — $m = 3$, the unit the cell actually reads.
4. **$\Omega = 4^3 = 64$ is the genetic code**: 61 codons for amino acids, 3 stop codons. The
   counting relationship from Part 3 lands exactly on a fact of biology.
5. **$H^{(m)}_{\max} = 6$ bits per codon** — three nucleotides at 2 bits each, the linear scaling
   of Part 3.
6. **Compare per character.** $3.75$ bits per codon is $1.25$ per nucleotide, below the $1.878$ of
   single-character reading.
7. **Sixteen codons cannot measure a 64-state alphabet.** The measured $3.75$ sits at 94% of the
   $\log_2 16 = 4$ bit sampling ceiling. Real estimates need thousands of codons.
