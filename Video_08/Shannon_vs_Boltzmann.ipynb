## 1. Where Part 7 left off

$$S_{\text{info}} = k_B \log \Omega_{\text{info}} = k_B \cdot N \cdot H(\mathcal{X}) \cdot \log 2$$

$$\Omega_{\text{info}} = b^{\,N \cdot H(\mathcal{X})} \qquad\xrightarrow{\;b=2\;}\qquad
\Omega_{\text{info}} = 2^{\,N \cdot H(\mathcal{X})}$$

The lecture restates this and then says plainly that the relation does **not** make the two
quantities identical — *"clearly the two entropies … they are not the same thing. We established
this last lecture. But they are related to each other by this relation."*

The rest of the lecture is spent earning that claim.

### Presenter's Notes — The Part 7 result, restated
![Board restating S info equals k B N H log 2 and omega info equals b to the N H](Screenshots/Part_8_00_05_45.jpg)

*Screenshot at timestamp 00:05:45*

---

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial, lgamma, log

K_B = 1.380649e-23          # J/K
LN2 = log(2)

def shannon_entropy(probs, base=2):
    p = np.asarray([q for q in probs if q > 0], dtype=float)
    return float(-np.sum(p * np.log(p) / np.log(base)))

def multinomial(N, counts):
    """Omega = N! / (n_1! ... n_L!) -- microstates of one macrostate."""
    assert sum(counts) == N, f"counts must sum to N (got {sum(counts)} vs {N})"
    denom = 1
    for n in counts:
        denom *= factorial(n)
    return factorial(N) // denom

def log2_multinomial(N, counts):
    """log2 of the same, without forming the (possibly enormous) integer."""
    lg = lambda k: lgamma(k + 1) / LN2
    return lg(N) - sum(lg(k) for k in counts)

print(f"k_B = {K_B:.6e} J/K")
```

## 2. Boltzmann's construction — phase space and cells

To describe $N$ identical particles completely you need, for each particle, **3 position
coordinates and 3 momentum components**. For $N$ particles that is a space of

$$6N \text{ dimensions}$$

called the **phase space**. Boltzmann's move was to partition this space into cells
$W_1, W_2, \ldots, W_L$ and count how many particles fall in each:

$$D = (n_1,\, n_2,\, \ldots,\, n_L), \qquad \sum_{i=1}^{L} n_i = N$$

### Three levels of state, not two

The lecture is careful about a hierarchy that is easy to collapse:

| Level | What it specifies | Example |
|---|---|---|
| **Thermodynamic macrostate** | bulk observables | $T,\, V,\, P,\, E,\, N$ |
| **Boltzmann macrostate** | a distribution over cells | $D = (n_1, \ldots, n_L)$ |
| **Microstate** | which particle is in which cell | one specific assignment |

*"The thermodynamic macrostate contains many Boltzmann macrostates, and each Boltzmann macrostate
contains many, many microstates."* At equilibrium the most probable Boltzmann macrostate is the
**Boltzmann distribution** — the one with the largest number of microstates.

### The count is the multinomial coefficient

For the $i$-th macrostate at energy $E$:

$$\boxed{\;\Omega_i(E) = \frac{N!}{n_1!\; n_2! \cdots n_L!}\;}$$

which is **exactly** the expression Part 6 derived for counting distinct messages. Same
combinatorics, arrived at independently from a physical problem.

### Presenter's Notes — The state hierarchy, $6N$ phase space, and $\Omega_i(E)$
![Board showing thermodynamic macrostate, Boltzmann macrostate, 6N phase space and the multinomial coefficient](Screenshots/Part_8_00_17_25.jpg)

*Screenshot at timestamp 00:17:25*

---

```python
# The same coefficient, computed for a "gas" and for a "message" with the same shape.
N = 12
distribution = [5, 4, 2, 1]        # n_1 .. n_L, summing to N

omega_gas = multinomial(N, distribution)
omega_msg = multinomial(N, distribution)

print("A Boltzmann macrostate: 12 particles across 4 phase-space cells")
print(f"  D = (n1..nL) = {tuple(distribution)}   ->  Omega = {omega_gas:,} microstates\n")
print("A message macrostate: 12 letters over a 4-letter alphabet")
print(f"  counts       = {tuple(distribution)}   ->  Omega = {omega_msg:,} messages\n")
assert omega_gas == omega_msg
print("Identical -- the combinatorics does not know or care which one it is counting.")
print("That shared formula is what the rest of this notebook has to interpret.")
```

## 3. Why systems go to equilibrium

With $\Omega$ in hand, the arrow of time follows from counting. A system *"evolves from a state of
lower probability to states of higher probability, and equilibrium corresponds to the state of the
highest probability"* — because the equilibrium macrostate simply has more microstates in it than
any other, so a randomly chosen microstate is overwhelmingly likely to belong to it.

Combining $\Omega_{\text{final}} \geq \Omega_{\text{initial}}$ with $S = k_B \ln \Omega$:

$$S_{\text{final}} \geq S_{\text{initial}} \qquad\Longrightarrow\qquad \boxed{\;dS \geq 0\;}$$

which is the second law of thermodynamics, obtained purely by counting states.

> **A caveat the lecture makes explicitly.** Equilibrium is an idealisation. *"In reality everything
> is non-equilibrium, but it's a good approximation."* A real system always contains some
> non-equilibrium macrostates; they are just typically very small.

> **Forward reference.** The lecturer flags that the **second law of infodynamics**, coming in later
> lectures, runs the *opposite* way: $dS_{\text{info}}$ decreases over time, and the two balance. He
> raises it here precisely so the contrast with $dS \geq 0$ is already in place.

### Presenter's Notes — $S = k_B \ln \Omega$ from $\partial S/\partial E = 1/T$, and $dS \geq 0$
![Board deriving S equals k B ln omega from the temperature relation and concluding dS greater than or equal to zero](Screenshots/Part_8_00_40_20.jpg)

*Screenshot at timestamp 00:40:20*

---

```python
# The second law by counting: let a partition come down and watch Omega grow.
# 20 particles, initially all confined to the left half (cells 0-1) of 4 cells.
N_p = 20
initial = [10, 10, 0, 0]           # constrained
final = [5, 5, 5, 5]               # free to spread over all four cells

for label, dist in [("initial (confined to 2 cells)", initial),
                    ("final   (spread over 4 cells)", final)]:
    w = multinomial(N_p, dist)
    S = K_B * np.log(float(w))
    print(f"{label}: D = {tuple(dist)}")
    print(f"    Omega = {w:>18,}     S = k_B ln(Omega) = {S:.4e} J/K")

w_i, w_f = multinomial(N_p, initial), multinomial(N_p, final)
dS = K_B * (np.log(float(w_f)) - np.log(float(w_i)))
print(f"\nOmega_final / Omega_initial = {w_f / w_i:,.1f}")
print(f"dS = {dS:.4e} J/K   ->  dS >= 0 : {dS >= 0}")
```

## 4. The comparison

Here is the lecture's answer, laid out as he presents it. Every row is a place where the two
constructions use the same symbol for a different thing.

| | **Boltzmann entropy** | **Shannon entropy** |
|---|---|---|
| Applies to | gas, particles, physical states, a crystal | a message, information states |
| $N$ | number of single-particle microstates | number of letters in a message |
| The list | $(n_1, n_2, \ldots, n_l)$ — cell distribution | $\{x_1, x_2, \ldots, x_l\}$ — an $l$-letter alphabet |
| $\{p_1,\ldots,p_l\}$ | probability distribution **over cells** | probability distribution **over letters** |
| $p_j = n_j/N$ | probability a microstate occurs in cell $j$ | probability $x_j$ occurs in a message |
| $N p_j$ | number of microstates in cell $w_j$ | number of $x_j$ letters in a message |
| Entropy | $S = k_B \ln \Omega = -N k_B \sum_{j=1}^{l} p_j \ln p_j$ | $H(\mathcal{X}) = -\sum_{j=1}^{l} p_j \log_b p_j$ |
| Units | J/K | bits |

The lecturer's summary: *"Similar to the Boltzmann, but it's not the same units, not quite the same.
And actually these quantities … refer to very different entities."*

## 5. Reading the last row carefully

The two entropy expressions in the final row of that table are worth putting directly beside each
other, because the difference between them *is* the Part 7 bridge:

$$S = -N k_B \sum_j p_j \ln p_j \qquad\qquad H(\mathcal{X}) = -\sum_j p_j \log_b p_j$$

Two differences, and both matter.

**Boltzmann carries an $N$; Shannon does not.** $S$ is **extensive** — double the system and you
double the entropy. $H$ is **intensive** — it is entropy *per character*, and doubling the message
leaves it unchanged. This is not a cosmetic difference in convention; it is a statement about what
kind of quantity each one is.

**Boltzmann carries a $k_B$ and reports J/K; Shannon is dimensionless and reports bits.** $k_B$ is
what converts a pure count into a thermodynamic quantity with units of energy per temperature.

Put those together and Part 7's relation is just the bookkeeping between them:

$$S = \underbrace{N}_{\text{make it extensive}} \cdot \underbrace{k_B \ln 2}_{\text{bits} \to \text{J/K}} \cdot \; H(\mathcal{X})$$

So $H$ is the per-character, dimensionless core; $S$ is that same core scaled by the size of the
system and dressed in physical units. They are not equal, they are not even the same *kind* of
quantity — and that is exactly why a conversion factor exists at all.

```python
probs = [0.4, 0.3, 0.2, 0.1]
H = shannon_entropy(probs, base=2)          # bits per character
H_nat = shannon_entropy(probs, base=np.e)   # nats per character

print(f"distribution p = {probs}")
print(f"H = {H:.6f} bits/character  =  {H_nat:.6f} nats/character\n")
print(f"{'N':>10}{'H (bits/ch)':>14}{'S = N k_B H ln2 (J/K)':>26}")
print("-" * 50)
for N_sys in (100, 1_000, 10_000, 100_000):
    S = N_sys * K_B * H * LN2
    print(f"{N_sys:>10,}{H:>14.6f}{S:>26.6e}")

print("\nH is unchanged by system size (intensive).")
print("S scales linearly with it (extensive).  The bridge is exactly N * k_B * ln2.\n")

# and confirm the two forms of the same number agree
N_chk = 5000
via_bits = N_chk * K_B * H * LN2
via_nats = N_chk * K_B * H_nat
print(f"N k_B H_bits ln2 = {via_bits:.10e} J/K")
print(f"N k_B H_nats     = {via_nats:.10e} J/K")
assert np.isclose(via_bits, via_nats)
print("identical -- ln2 is only the bits-to-nats conversion.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))
sizes = np.array([100, 500, 1000, 5000, 10000, 50000, 100000])

axes[0].semilogx(sizes, [H] * len(sizes), 'o-', color='steelblue', linewidth=2, markersize=6)
axes[0].set_xlabel('system size $N$', fontsize=11)
axes[0].set_ylabel('$H(\\mathcal{X})$  (bits per character)', fontsize=11)
axes[0].set_title('Shannon: intensive\nflat in $N$', fontsize=12)
axes[0].set_ylim(0, H * 1.6)
axes[0].grid(alpha=0.3)

axes[1].loglog(sizes, sizes * K_B * H * LN2, 'o-', color='crimson', linewidth=2, markersize=6)
axes[1].set_xlabel('system size $N$', fontsize=11)
axes[1].set_ylabel('$S$  (J/K)', fontsize=11)
axes[1].set_title('Boltzmann: extensive\nlinear in $N$', fontsize=12)
axes[1].grid(alpha=0.3, which='both')

plt.tight_layout()
plt.show()

print("Same underlying distribution, same -sum p log p core -- but one is a rate")
print("and the other is a total. That is the distinction the lecture is drawing.")
```

## 6. The disagreement, stated fairly

The lecture frames this comparison as a direct reply. It reports that physicist **Sabine
Hossenfelder** *"claimed that there is no difference between Shannon entropy and Boltzmann
entropy"*, and offers the table as a rebuttal.

> **A note on sourcing.** The characterisation of her position above is the lecture's, quoted as
> given. This notebook has not consulted her original argument, and readers who want to weigh the
> exchange should go to both sources directly rather than rely on either summary.

What can be said without taking a side is that the two claims are not addressing quite the same
question, and the table itself shows why:

**The formalism is genuinely shared.** Both quantities are $-\sum_j p_j \log p_j$ evaluated on a
probability distribution, and both microstate counts are the same multinomial coefficient. The table
does not dispute this — it reproduces the same algebra in both columns. Anyone saying the two are
"the same" in the sense of *same mathematical object* has that much of a point.

**The referents are genuinely different.** $N$ counts particles on one side and letters on the
other; $p_j$ is an occupancy over phase-space cells on one side and a symbol frequency on the other;
one is extensive with units of J/K and the other intensive and dimensionless. Anyone saying they
are "not the same" in the sense of *same physical quantity* has that much of a point too.

So the substantive question — the one the algebra cannot settle — is whether a shared mathematical
structure between a physical system and an abstract message licenses treating information as
physical. That is what Part 6's postulate asserted, and it is what the remaining lectures on
mass–energy–information equivalence set out to defend.

> This notebook does not resolve that, and it would be overreaching to pretend the derivation does.
> [Part 7, Section 8](../Video_07/Shannon_and_Boltzmann.ipynb) raised the same point and reached the
> same limit.

## Summary

| | Boltzmann | Shannon |
|---|---|---|
| Object | physical system | message |
| State space | $6N$-dim phase space, cells $W_1..W_L$ | alphabet $x_1..x_L$ |
| Macrostate | $(n_1,\ldots,n_L)$ over cells | $(n_1,\ldots,n_L)$ over letters |
| Microstate count | $\Omega = N!/\prod_j n_j!$ | $\Omega = N!/\prod_j n_j!$ |
| Entropy | $S = -Nk_B\sum_j p_j \ln p_j$ | $H = -\sum_j p_j \log_b p_j$ |
| Character | extensive, J/K | intensive, bits |

and the bridge between the two columns, from Part 7:

$$S = N \, k_B \ln 2 \cdot H(\mathcal{X})$$

### Key takeaways

1. **Boltzmann's microstate count is the same multinomial coefficient** as Part 6's message count —
   arrived at independently, from partitioning a $6N$-dimensional phase space into cells.
2. **There are three levels of state**, not two: a thermodynamic macrostate contains many Boltzmann
   macrostates, each of which contains many microstates.
3. **The second law follows from counting.** $\Omega_{\text{final}} \geq \Omega_{\text{initial}}$
   with $S = k_B\ln\Omega$ gives $dS \geq 0$ directly.
4. **Equilibrium is an idealisation** — real systems always retain small non-equilibrium macrostates.
5. **Every symbol in the shared formula refers to something different** — particles vs letters,
   cells vs symbols, occupancy vs frequency.
6. **$S$ is extensive and dimensional; $H$ is intensive and dimensionless.** The Part 7 conversion
   factor $N k_B \ln 2$ is precisely what reconciles those two facts.
7. **The lecturer's verdict**: connected by a derived relation, but not the same thing.

### Where this leads

With the distinction drawn, the series turns to information physics proper — the
**mass–energy–information equivalence principle**, the **second law of information dynamics** (where
$dS_{\text{info}}$ runs opposite to $dS$), and the **information catastrophe**.
