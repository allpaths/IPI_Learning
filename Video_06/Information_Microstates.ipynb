# The Information Microstates
## Counting the distinct messages a set of characters can form

> 📺 **[Watch the original lecture on YouTube](https://youtu.be/gaSykBTbfsU)**

Since [Part 3](../Video_03/Information_Entropy_of_m_blocks.ipynb) the symbol $\Omega$ has been
carrying a promise. It was introduced as the count of distinct $m$-blocks, but chosen deliberately
— it is Boltzmann's symbol for the number of **microstates** of a physical system, and the lecture
said at the time that the connection would be made later.

This is later.

The question here is concrete: given a message of $N$ characters drawn from $U$ distinct ones, **in
how many distinct ways could those characters be arranged?** The lecture postulates that this count
*is* the number of information microstates, in direct analogy with the microstates of a
thermodynamic system, and then computes it two ways — with the **multinomial coefficient** for one
fixed letter composition, and with the **multinomial theorem** for all compositions at once.

In the lecturer's words, this is *"super essential for what comes next"* — the mass–energy–information
equivalence principle and the second law of infodynamics both rest on it.

## 1. Macrostates, microstates, and Boltzmann

A **macrostate** specifies a system by its macroscopic properties — pressure, temperature, volume,
mass. A **microstate** is one specific arrangement of the system's components consistent with that
macrostate.

The question that defines the count is:

> Given a system of many components, in how many distinguishable ways could we arrange those
> components so that the system still looks macroscopically the same?

That number is $\Omega$, the number of microstates compatible with the macrostate. Boltzmann's
result — the bridge from thermodynamics to statistical mechanics — links it to entropy:

$$\boxed{\;S = k_B \log \Omega\;}$$

where $k_B$ is Boltzmann's constant, of order $10^{-23}$ J/K.

> **On the logarithm base.** The lecture writes $\log$ rather than $\ln$ deliberately. In physics
> this relation is conventionally natural-log, giving $S$ in J/K. But the base is a choice of units,
> and Section 9 is where leaving it open pays off — in base 2 the same expression counts *bits*.

A note the lecture is careful about: thermodynamic entropy and Boltzmann entropy are distinct
definitions that coincide for systems with large numbers of particles.

## 2. The information system

Set up the message exactly as in Parts 1–5. A set of $N$ characters, of which $U$ are distinct,
with $N \geq U$:

$$\mathcal{X} = \{x_1,\, x_2,\, \ldots,\, x_U\}$$

carrying a probability distribution

$$P = \{p_1,\, p_2,\, \ldots,\, p_U\}, \qquad p_j \geq 0, \qquad \sum_{j=1}^{U} p_j = 1$$

where $p_j$ is the probability that character $x_j$ occurs.

The step this lecture needs is turning probabilities into **counts**. If character $x_j$ occurs with
probability $p_j$ in a message of $N$ characters, the number of times it actually appears is

$$n_j = p_j N$$

so $n_1 = p_1 N$ occurrences of $x_1$, $n_2 = p_2 N$ of $x_2$, and so on, subject to

$$\sum_{j=1}^{U} n_j = N$$

The vector $(n_1, n_2, \ldots, n_U)$ is the **composition** of the message — how many of each
character it contains, saying nothing about their order. That distinction is the whole lecture:
composition is the macrostate, arrangement is the microstate.

```python
import numpy as np
import matplotlib.pyplot as plt
from math import factorial
from collections import Counter
from itertools import product

def composition(message):
    """The (n_1, ..., n_U) counts of a message -- its macrostate."""
    return Counter(message)

message = "AAB"
N = len(message)
alphabet = sorted(set(message))
U = len(alphabet)
counts = composition(message)

print(f"message   = {message}")
print(f"N         = {N}   (total characters)")
print(f"U         = {U}   (distinct characters)")
print(f"X         = {{{', '.join(alphabet)}}}\n")
for ch in alphabet:
    print(f"  n_{ch} = {counts[ch]},  p_{ch} = {counts[ch]}/{N} = {counts[ch] / N:.4f}")
print(f"\nsum of n_j = {sum(counts.values())} = N  ✓")
```

## 3. The postulate — distinct messages *are* microstates

Here is the conceptual move the lecture makes explicit, and it is worth pausing on because
everything downstream depends on it:

> The number of distinct messages that can be formed from the same set of letters is $\Omega$, and
> **this is the number of microstates compatible with the macrostate.**

The macrostate is the composition — *this message contains two A's and one B*. The microstates are
the specific orderings — $\texttt{AAB}$, $\texttt{ABA}$, $\texttt{BAA}$ — each carrying the same
letters in a different arrangement. A message has $\Omega$ ways of being transmitted using the same
set of letters.

That is a postulate, not a derivation: it asserts that arrangements of a message are the
informational analogue of arrangements of particles. Its justification is what it buys — once
$\Omega$ is a count of message arrangements, Boltzmann's $S = k_B \log \Omega$ applies to messages,
and information theory and thermodynamics are speaking about the same quantity.

---
## 4. Counting one macrostate — the multinomial coefficient

For a *fixed* composition, the count comes from combinatorics. The **multinomial coefficient** gives
the number of distinct arrangements of $N$ items of which $n_1$ are of one kind, $n_2$ of another,
and so on:

$$\Omega = \frac{N!}{(p_1 N)!\,(p_2 N)!\cdots(p_U N)!}$$

or written directly in terms of the occurrence counts,

$$\boxed{\;\Omega = \frac{N!}{n_1!\; n_2! \cdots n_U!}\;}
\qquad \text{subject to} \qquad \sum_{j=1}^{U} n_j = N$$

The intuition: there are $N!$ ways to order $N$ distinguishable items, but the $n_1$ copies of $x_1$
are *not* distinguishable from one another, so every arrangement has been counted $n_1!$ times over
— and likewise for each other character. Dividing by $n_1!\,n_2!\cdots n_U!$ removes exactly that
overcounting.

### Presenter's Notes — The multinomial coefficient and the multinomial theorem
![Board showing the multinomial coefficient in both forms, the constraint, and the multinomial theorem giving U to the N](Screenshots/Part_6_00_16_10.jpg)

*Screenshot at timestamp 00:16:10*

```python
def multinomial(N, counts):
    """Omega = N! / (n_1! n_2! ... n_U!) -- microstates for one fixed composition."""
    assert sum(counts) == N, f"counts must sum to N (got {sum(counts)} vs {N})"
    denom = 1
    for n in counts:
        denom *= factorial(n)
    return factorial(N) // denom

def microstates(counts_map):
    """Every distinct arrangement of a given composition, in lexicographic order."""
    pool = ''.join(ch * n for ch, n in sorted(counts_map.items()))
    return sorted(set(''.join(p) for p in _perms(pool)))

def _perms(s):
    if len(s) <= 1:
        yield s
        return
    seen = set()
    for i, ch in enumerate(s):
        if ch in seen:
            continue
        seen.add(ch)
        for rest in _perms(s[:i] + s[i + 1:]):
            yield ch + rest

n_vec = [counts[ch] for ch in alphabet]
Omega = multinomial(N, n_vec)
states = microstates(counts)

print(f"Omega = {N}! / ({' * '.join(f'{n}!' for n in n_vec)}) "
      f"= {factorial(N)} / {np.prod([factorial(n) for n in n_vec])} = {Omega}")
print(f"\nthe {Omega} microstates: {{{'; '.join(states)}}}")
assert len(states) == Omega, "enumeration disagrees with the formula"
print("\nenumeration matches the multinomial coefficient.")
```

## 5. The worked example — $\texttt{AAB}$

The lecture takes a deliberately tiny system so the whole thing can be done by hand:

$$\text{message} = \texttt{AAB} \qquad N = 3, \quad U = 2, \quad \mathcal{X} = \{A, B\}$$
$$n_A = 2, \qquad n_B = 1 \qquad\Longrightarrow\qquad p_A = \tfrac{2}{3}, \quad p_B = \tfrac{1}{3}$$

Applying the coefficient:

$$\Omega = \frac{N!}{n_A!\; n_B!} = \frac{3!}{2!\;1!} = \frac{6}{2} = 3$$

and the three microstates are exactly

$$\{\texttt{AAB},\; \texttt{ABA},\; \texttt{BAA}\}$$

There is no fourth way to arrange two A's and one B. The coefficient and the enumeration agree.

## 6. Counting *every* macrostate — the multinomial theorem

The coefficient answers a question about one composition. The lecture then asks the larger one:

> Given $N$ letters and $U$ distinct characters, how many messages could be formed **taking into
> account all possible compositions** — not just one set of occurrences, but all of them?

Summing the coefficient over every composition satisfying $\sum_j n_j = N$ gives the
**multinomial theorem**:

$$\Omega_{\text{tot}} = \sum_{n_1 + n_2 + \cdots + n_U = N} \frac{N!}{n_1!\;n_2!\cdots n_U!}$$

and that messy-looking sum collapses to something strikingly simple:

$$\boxed{\;\Omega_{\text{tot}} = U^{\,N}\;}$$

The lecture's own verdict on the shortcut: *"instead of doing all this messy stuff, you can just do
$U$ to power $N$ and you get a free lunch."*

The direct argument for why: building a message means choosing one of $U$ characters for each of
$N$ positions independently, which is $U \times U \times \cdots \times U = U^N$. The sum over
compositions must give the same total, because sorting all $U^N$ messages into bins by composition
and adding up the bin sizes cannot change how many there are.

```python
def omega_total_by_sum(N, U):
    """Sum the multinomial coefficient over every composition -- the long way."""
    total = 0
    rows = []
    for comp in _compositions(N, U):
        w = multinomial(N, comp)
        rows.append((comp, w))
        total += w
    return total, rows

def _compositions(N, U):
    """All (n_1, ..., n_U) with non-negative entries summing to N."""
    if U == 1:
        yield (N,)
        return
    for first in range(N + 1):
        for rest in _compositions(N - first, U - 1):
            yield (first,) + rest

total_sum, rows = omega_total_by_sum(N, U)
total_pow = U ** N

print(f"the long way  : sum of multinomial coefficients = {total_sum}")
print(f"the shortcut  : U^N = {U}^{N} = {total_pow}")
assert total_sum == total_pow
print("\nthey agree.\n")

# every message of length N over the alphabet
all_msgs = [''.join(p) for p in product(alphabet, repeat=N)]
print(f"all {len(all_msgs)} messages: {{{', '.join(all_msgs)}}}")
assert len(all_msgs) == total_pow
```

## 7. Both routes, side by side

The lecture verifies the theorem by tabulating every composition of a 3-character binary message.
With $n_A + n_B = N = 3$ there are just four:

| $n_A$ | $n_B$ | microstates | $\dfrac{N!}{n_A!\,n_B!}$ |
|---|---|---|---|
| 0 | 3 | $\texttt{BBB}$ | $\dfrac{3!}{0!\,3!} = 1$ |
| 1 | 2 | $\texttt{ABB},\ \texttt{BAB},\ \texttt{BBA}$ | $\dfrac{3!}{1!\,2!} = 3$ |
| 2 | 1 | $\texttt{AAB},\ \texttt{ABA},\ \texttt{BAA}$ | $\dfrac{3!}{2!\,1!} = 3$ |
| 3 | 0 | $\texttt{AAA}$ | $\dfrac{3!}{3!\,0!} = 1$ |
| | | | $\sum = 1+3+3+1 = \mathbf{8}$ |

and $U^N = 2^3 = 8$. The two routes land on the same number, and the microstates listed in the
table are precisely the eight messages the shortcut counts.

Note $0! = 1$ is doing quiet work in the first and last rows — a composition with zero occurrences
of a character contributes exactly one arrangement, which is right: if there are no A's, the only
possible message is all B's.

## 8. $\Omega_{\text{tot}} = U^N$ is Part 3's formula again

Worth noticing, because the series has now written the same expression twice for different reasons.

[Part 3](../Video_03/Information_Entropy_of_m_blocks.ipynb) derived $\Omega = U^{\,m}$ as the number
of distinct **$m$-blocks** available over an alphabet of $U$ characters. Here $\Omega_{\text{tot}} =
U^{\,N}$ is the number of distinct **messages** of length $N$.

These are the same statement. A message of $N$ characters *is* an $N$-block — so setting $m = N$ in
Part 3's relationship gives Part 6's. The whole message is simply the largest block you can take,
which is precisely the case Part 3 flagged when it listed $m = N$ as one end of the range of block
sizes.

The entropy consequence carries over intact. Part 3 had $H_{\max} = \log_2 \Omega$; applying it here:

$$\log_2 \Omega_{\text{tot}} = \log_2 U^{\,N} = N \log_2 U = N \cdot H(\mathcal{X})_{\max}$$

So the logarithm of the *total* microstate count is the maximum information the message could carry
— $N$ characters at $\log_2 U$ bits each. Boltzmann's $\log \Omega$, read in base 2, is already
measuring bits.

```python
H_max = np.log2(U)
print(f"log2(Omega_tot) = log2({U}^{N}) = {np.log2(total_pow):.4f} bits")
print(f"N * H_max       = {N} x log2({U}) = {N * H_max:.4f} bits")
assert np.isclose(np.log2(total_pow), N * H_max)
print("\nidentical -- Part 3's H_max = log2(Omega), now at whole-message scale.")

print(f"\nPart 3:  Omega = U^m  (distinct m-blocks)")
print(f"Part 6:  Omega_tot = U^N  (distinct N-character messages)")
print(f"         -- the same relationship at m = N")
```

## 9. Where this is heading — $\log_2 \Omega \approx N \cdot H(\mathcal{X})$

> **Confirmed by Part 7.** Part 6 assembles the machinery and states that it *"will allow the
> creation of a link between information theory and physics"*, but stops before making that link
> explicit. This section took the one further step ahead of the lecture — and
> [Part 7](../Video_07/Shannon_and_Boltzmann.ipynb) is the lecturer's own treatment of it,
> reaching the same result by the same route (Stirling applied to the multinomial coefficient).
> He carries a general base $b$ throughout where this section fixes natural log, which amounts to
> the same statement: $\log_b \Omega = N\,H_b$. See Part 7 for his derivation, the inverted form
> $\Omega_{\text{info}} = b^{\,N H}$, and an erratum the base-generality surfaced.

Take the multinomial coefficient for a message whose composition follows the distribution
$\{p_j\}$, so $n_j = p_j N$, and ask how $\log \Omega$ behaves for large $N$. Using Stirling's
approximation $\ln n! \approx n \ln n - n$:

$$\ln \Omega = \ln N! - \sum_j \ln n_j! \approx \left(N \ln N - N\right) - \sum_j \left(n_j \ln n_j - n_j\right)$$

The $-N$ and $+\sum_j n_j = +N$ cancel. Substituting $n_j = p_j N$:

$$\ln \Omega \approx N \ln N - \sum_j p_j N \ln (p_j N)
= N \ln N - N\sum_j p_j \ln p_j - N \ln N \underbrace{\sum_j p_j}_{=\,1}$$

The two $N \ln N$ terms cancel, leaving

$$\boxed{\;\ln \Omega \approx -N \sum_j p_j \ln p_j = N \cdot H(\mathcal{X})\;}$$

or in base 2, $\log_2 \Omega \approx N \cdot H(\mathcal{X})$ with $H$ in bits per character.

**This is the bridge.** Boltzmann counts microstates and takes a logarithm; Shannon sums
$-p\log p$. For a message of $N$ characters they are the same quantity — Shannon entropy is the
per-character version of $\log \Omega$. Substituting into Boltzmann's relation:

$$S = k_B \ln \Omega \approx k_B N \cdot H(\mathcal{X})$$

The thermodynamic entropy of a system and the Shannon information entropy of a message describing
it differ only by $k_B N$ — a change of units and a count of characters.

The approximation is Stirling's, so it is asymptotic in $N$. The cell below shows how fast it
converges.

```python
from math import lgamma, log as _ln

LN2 = _ln(2)

def log2_factorial(k):
    """log2(k!) via lgamma.

    Computing multinomial() directly and taking log2 of the result works only
    for small N: the coefficient is an exact Python int that quickly exceeds
    the float64 range (N = 1000 already puts it past 10^400), at which point
    numpy cannot convert it at all. Working in log space throughout avoids
    ever forming that integer.
    """
    return lgamma(k + 1) / LN2

def log2_omega(N, probs):
    """log2 of the multinomial coefficient for the composition round(p*N)."""
    n = [int(round(p * N)) for p in probs]
    if sum(n) != N:            # fix rounding drift on the largest bin
        n[int(np.argmax(n))] += N - sum(n)
    return log2_factorial(N) - sum(log2_factorial(k) for k in n), n

# Verify the log-space form against exact integer arithmetic, in the range
# where the exact integer still fits a float.
for Nv in (10, 50, 100, 200):
    approx, n_chk = log2_omega(Nv, [0.2, 0.3, 0.5])
    exact = float(np.log2(float(multinomial(Nv, n_chk))))
    assert abs(approx - exact) < 1e-9, (Nv, approx, exact)
print("lgamma form agrees with exact integer arithmetic to 1e-9 for N <= 200.\n")

probs = [0.2, 0.3, 0.5]
H = -sum(p * np.log2(p) for p in probs)
print(f"distribution p = {probs}")
print(f"Shannon H(X)   = {H:.6f} bits per character\n")
print(f"{'N':>7}{'log2(Omega)':>16}{'N*H':>14}{'ratio':>10}{'per-char err':>15}")
print("-" * 62)
for Nv in [10, 50, 100, 500, 1000, 5000, 20000]:
    lo, _ = log2_omega(Nv, probs)
    print(f"{Nv:>7}{lo:>16.3f}{Nv * H:>14.3f}{lo / (Nv * H):>10.4f}{abs(lo / Nv - H):>15.6f}")
print(f"\nlog2(Omega)/N converges to H = {H:.6f} bits per character.")
```

```python
Ns_plot = np.unique(np.logspace(1, 4.3, 40).astype(int))
per_char = [log2_omega(int(nv), probs)[0] / nv for nv in Ns_plot]
H_max_3 = np.log2(len(probs))

fig, ax = plt.subplots(figsize=(9, 5))
ax.semilogx(Ns_plot, per_char, 'o-', color='steelblue', linewidth=2, markersize=4,
            label='$\\log_2\\Omega / N$')
ax.axhline(H, color='crimson', linestyle='--', linewidth=1.5,
           label=f'Shannon $H = {H:.4f}$ bits')
ax.axhline(H_max_3, color='gray', linestyle=':', linewidth=1.5,
           label=f'$H_{{max}} = {H_max_3:.4f}$ bits')
ax.set_xlabel('message length $N$', fontsize=11)
ax.set_ylabel('bits per character', fontsize=11)
ax.set_title("Boltzmann's count and Shannon's sum are the same quantity\n"
             '$\\log_2\\Omega / N \\to H$ as $N$ grows', fontsize=12)
ax.legend(fontsize=10)
ax.grid(alpha=0.3)
plt.tight_layout()
plt.show()

err_small, err_big = abs(per_char[0] - H), abs(per_char[-1] - H)
print(f"at N = {Ns_plot[0]:,}: off by {err_small:.4f} bits/character")
print(f"at N = {Ns_plot[-1]:,}: off by {err_big:.4f} bits/character "
      f"({err_big / err_small:.1%} of the small-N error)")
```

## Summary

| Quantity | Definition | Meaning |
|---|---|---|
| Macrostate | the composition $(n_1,\ldots,n_U)$ | how many of each character |
| Microstate | one arrangement of those characters | one distinct message |
| Microstates of one macrostate | $\Omega = \dfrac{N!}{n_1!\,n_2!\cdots n_U!}$ | multinomial coefficient |
| Constraint | $\sum_j n_j = N$, with $n_j = p_j N$ | composition must fill the message |
| Microstates of all macrostates | $\Omega_{\text{tot}} = \sum \dfrac{N!}{n_1!\cdots n_U!} = U^{\,N}$ | multinomial theorem |
| Boltzmann | $S = k_B \log \Omega$ | entropy from a state count |
| The bridge | $\log_2 \Omega \approx N \cdot H(\mathcal{X})$ | Shannon entropy is $\log\Omega$ per character |

### Key takeaways

1. **A microstate of a message is an arrangement of its characters.** The composition is the
   macrostate; the orderings consistent with it are the microstates.
2. **The multinomial coefficient counts one macrostate.** $N!$ orderings, divided by $n_j!$ for
   each character to remove the overcounting of indistinguishable copies.
3. **The multinomial theorem counts them all**, and collapses to $\Omega_{\text{tot}} = U^{\,N}$ —
   the "free lunch" shortcut over summing coefficients.
4. **Both routes agree.** For $\texttt{AAB}$: $1+3+3+1 = 8 = 2^3$.
5. **$U^N$ is Part 3's $U^m$ at $m = N$.** A message is just the largest block you can take, so
   $\log_2 \Omega_{\text{tot}} = N \cdot H_{\max}$ — the same relationship, at whole-message scale.
6. **Stirling closes the loop**: $\log_2 \Omega \approx N \cdot H(\mathcal{X})$. Boltzmann's
   logarithm of a state count and Shannon's $-\sum p\log p$ are the same quantity, differing only by
   the number of characters.

### Where this leads

With $\Omega$ established as a count of *information* microstates, $S = k_B \log \Omega$ can be
applied to a message rather than a gas. The lecturer names what that unlocks: the
**mass–energy–information equivalence principle** and the **second law of information dynamics** —
both, in his words, emanating from these principles.
