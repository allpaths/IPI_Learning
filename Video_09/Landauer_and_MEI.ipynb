## 1. Landauer's principle

In **1961**, Rolf Landauer proposed that a bit of information is a *physical* entity — that
information cannot be divorced from the hardware carrying it, and that logically irreversible
operations therefore carry a thermodynamic cost.

The key distinction is between operations that are **logically reversible** and those that are not.
Copying a bit, or a NOT gate, throws nothing away — you can run them backwards. **Erasure** cannot
be run backwards: whether the bit was 0 or 1, afterwards it is in a known reset state, and the
information about which it *was* is gone. That loss is what has to be paid for.

The lecture's phrasing: erase operations are **dissipative**, and Landauer quantified how much they
must dissipate.

## 2. The bound, from entropy accounting

The derivation is a direct application of the second law, using exactly the entropy bookkeeping
Parts 6–8 established.

Erasing one bit removes one bit of information entropy. In thermodynamic units that is

$$\Delta S_{\text{info}} = -k_B \ln 2$$

The second law applies to the **total** entropy, information plus physical:

$$\Delta S_{\text{tot}} = \Delta S_{\text{phys}} + \Delta S_{\text{info}} \geq 0$$

Substituting:

$$\Delta S_{\text{phys}} - k_B \ln 2 \geq 0
\qquad\Longrightarrow\qquad
\Delta S_{\text{phys}} \geq k_B \ln 2$$

The physical entropy has to *increase* by at least as much as the information entropy fell. That
increase is heat dumped into the environment at temperature $T$, and $\Delta Q / T = \Delta S_{\text{phys}}$, so

$$\boxed{\;\Delta Q \geq k_B T \ln 2\;}$$

This is the **Landauer bound**: the minimum energy dissipated when one bit is irreversibly erased.
It is the point at which information acquires a price tag denominated in joules.

> Note the structure — the bound is about the **cost of a process** (erasure), not a property of a
> bit sitting undisturbed. Section 5 returns to this, because it is exactly where the next step
> does its work.

```python
import numpy as np
import matplotlib.pyplot as plt

K_B = 1.380649e-23           # J/K          (exact, SI definition)
C = 2.99792458e8             # m/s          (exact, SI definition)
LN2 = np.log(2)
EV = 1.602176634e-19         # J per eV     (exact, SI definition)
M_ELECTRON = 9.1093837015e-31  # kg

def landauer_energy(T):
    """Minimum energy dissipated erasing one bit at temperature T, in joules."""
    return K_B * T * LN2

print(f"{'T (K)':>10}{'Landauer bound (J)':>24}{'in eV':>14}")
print("-" * 48)
for T in (1, 77, 300, 1000):
    E = landauer_energy(T)
    label = {1: "  (near absolute zero)", 77: "  (liquid nitrogen)",
             300: "  (room temperature)", 1000: ""}[T]
    print(f"{T:>10}{E:>24.4e}{E / EV:>14.6f}{label}")

E300 = landauer_energy(300)
print(f"\nAt room temperature one erasure costs {E300:.4e} J = {E300 / EV:.4f} eV.")
print("Small -- but strictly greater than zero, and experimentally confirmed.")
```

## 3. From energy to mass — the M/E/I principle

Here is the step that gives the lecture its name, and the lecturer presents its logic as a chain of
equalities — literally writing $x = y$, $y = z \Rightarrow x = y = z$ on the board:

| Link | Relation | Due to |
|---|---|---|
| mass ↔ energy | $E = mc^2$ | Einstein, special relativity |
| energy ↔ information | $E = k_B T \ln 2$ per bit | Landauer |
| **mass ↔ information** | **?** | the proposed principle |

The physical picture offered: a bit at rest, storing information indefinitely, *"does so because it
acquires a bit of mass, by converting the energy contained in the bit into a small mass. If you
erase that information, that small mass gets reconverted back into energy and is dissipated as
heat"* — at exactly the Landauer value.

Setting the bit's rest energy equal to its erasure cost:

$$m_{\text{bit}} c^2 = k_B T \ln 2$$

$$\boxed{\;m_{\text{bit}} = \frac{k_B T \ln 2}{c^2}\;}$$

This is the **mass–energy–information equivalence principle**, and it is the lecturer's own
contribution rather than textbook physics. Section 5 is explicit about that.

### Presenter's Notes — $m_{\text{bit}}c^2 = k_B T \ln 2$ and the mass of a bit
![Board showing the boxed Landauer results leading to the mass of a bit formula](Screenshots/Part_9_00_38_20.jpg)

*Screenshot at timestamp 00:38:20*

---

```python
def mass_of_bit(T):
    """Proposed rest mass of one stored bit at temperature T, in kg."""
    return K_B * T * LN2 / C ** 2

m300 = mass_of_bit(300)
print(f"At T = 300 K:  m_bit = k_B T ln2 / c^2 = {m300:.4e} kg")
print(f"lecture states approximately 3.19e-38 kg -> match: "
      f"{np.isclose(m300, 3.19e-38, rtol=1e-2)}\n")

print(f"electron mass          = {M_ELECTRON:.4e} kg")
print(f"m_bit / m_electron     = {m300 / M_ELECTRON:.3e}")
print(f"-> a bit is about {M_ELECTRON / m300:.2e} times lighter than an electron.\n")

print(f"{'T (K)':>10}{'m_bit (kg)':>18}")
print("-" * 28)
for T in (0, 1, 77, 300, 1000):
    print(f"{T:>10}{mass_of_bit(T):>18.4e}")

print("\nAt T = 0 the mass vanishes. The lecture reads this as information being")
print("unable to exist at absolute zero -- consistent with the third law, which")
print("puts T = 0 out of reach anyway.")
```

```python
# Scale: how much information would it take to weigh something?
bits_per_kg = 1 / m300
print(f"bits needed to total 1 kg at 300 K : {bits_per_kg:.3e}")
print(f"                         ... bytes : {bits_per_kg / 8:.3e}\n")

print(f"{'stored data':<28}{'bits':>16}{'proposed mass (kg)':>22}")
print("-" * 66)
for label, nbits in [
    ("1 KB", 8 * 1024),
    ("1 GB", 8 * 1024 ** 3),
    ("1 TB hard drive", 8 * 1024 ** 4),
    ("human genome (2 bits/base)", 2 * int(3.1e9)),
]:
    print(f"{label:<28}{nbits:>16,}{nbits * m300:>22.3e}")

print("\nA terabyte would weigh about the mass of a few hundred thousand atomic nuclei --")
print("far below anything weighable. The lecture is candid that experimental")
print("confirmation is future work, not a present capability.")
```

```python
fig, axes = plt.subplots(1, 2, figsize=(13, 5))

T = np.linspace(0, 1000, 400)
axes[0].plot(T, landauer_energy(T) / EV, color='crimson', linewidth=2)
axes[0].axvline(300, color='gray', linestyle=':', linewidth=1.5, label='300 K')
axes[0].plot([300], [landauer_energy(300) / EV], 'o', color='crimson', markersize=7)
axes[0].set_xlabel('temperature $T$ (K)', fontsize=11)
axes[0].set_ylabel('$k_B T \\ln 2$  (eV per bit)', fontsize=11)
axes[0].set_title("Landauer bound\nestablished physics", fontsize=12)
axes[0].legend(fontsize=10)
axes[0].grid(alpha=0.3)

axes[1].plot(T, mass_of_bit(T), color='steelblue', linewidth=2)
axes[1].axvline(300, color='gray', linestyle=':', linewidth=1.5, label='300 K')
axes[1].plot([300], [m300], 'o', color='steelblue', markersize=7)
axes[1].set_xlabel('temperature $T$ (K)', fontsize=11)
axes[1].set_ylabel('$m_{bit}$  (kg)', fontsize=11)
axes[1].set_title("Proposed mass of a bit\nan open research proposal", fontsize=12)
axes[1].legend(fontsize=10)
axes[1].grid(alpha=0.3)

plt.tight_layout()
plt.show()

print("Both are linear in T and differ only by the constant 1/c^2 -- the plots are")
print("the same shape. Their scientific standing is not the same, which is the")
print("subject of section 5.")
```

## 5. What follows from what

The three claims are not equally secure, and the difference is worth being precise about — not to
diminish the proposal, but because a study companion should tell a reader which parts are safe to
rely on.

### Landauer's principle — established

Landauer's bound is mainstream statistical mechanics. It has been **experimentally verified**: a
2012 measurement on a single colloidal particle in an optical trap, and several experiments since,
have confirmed the $k_B T \ln 2$ floor. The lecture references these tests. Nothing in Section 2 is
controversial.

### The M/E/I principle — a proposal, not a deduction

The algebra in Section 3 is trivial: divide by $c^2$. The physical content sits entirely in the
equation it starts from,

$$m_{\text{bit}} c^2 = k_B T \ln 2$$

and that is an **assumption**, not something Landauer's principle gives you. The reason is a
mismatch the lecture's own derivation makes visible: $k_B T \ln 2$ is the heat dissipated **during
an erasure process**, and it depends on the temperature of the environment the erasure happens in.
A rest mass is a property of the object **while it sits there**, independent of any process. Reading
a process cost as a rest energy is the step that carries the claim, and it is the lecturer's own —
it does not follow from Landauer, and it is not part of the standard reading of Landauer.

That is precisely the pattern flagged in [Part 6](../Video_06/Information_Microstates.ipynb), where
identifying message arrangements with physical microstates was likewise a postulate rather than a
derivation, and in [Part 8](../Video_08/Shannon_vs_Boltzmann.ipynb), where the shared formalism did
not by itself establish shared physics. Each is a place the series moves from mathematics to a
physical claim.

None of this makes the principle wrong. It makes it a **testable hypothesis** — which is how the
lecturer presents it, and he is candid that the predicted masses are far below current measurement
capability.

### The fifth state of matter — speculative

Explicitly framed as speculation by the lecturer himself (*"it might just be…"*), and dependent on
M/E/I being correct in the first place.

> **How to read this notebook.** Section 2 is textbook. Section 3 is a published research proposal
> by the lecturer. Section 4 is frontier speculation built on it. All three are worth understanding;
> only the first should be cited as settled.

```python
# The dependency chain, made explicit.
claims = [
    ("Landauer bound      dQ >= k_B T ln2", "established",
     "second law + information entropy; experimentally confirmed since 2012"),
    ("M/E/I  m_bit = k_B T ln2 / c^2", "proposed",
     "requires assuming a bit's REST energy equals its ERASURE cost"),
    ("Information as a fifth state", "speculative",
     "requires M/E/I first; presented by the lecturer as an open direction"),
]

print(f"{'claim':<40}{'status':<14}{'rests on'}")
print("-" * 108)
for claim, status, rests in claims:
    print(f"{claim:<40}{status:<14}{rests}")

print("\nEach row depends on the one above it. The algebra connecting them is sound;")
print("what varies is the strength of the physical assumption each step introduces.")
```

### Key takeaways

1. **Erasure is the dissipative operation.** Logically reversible operations carry no such floor;
   erasure destroys the record of which state the bit was in, and that is what costs.
2. **The bound follows from the second law applied to total entropy.** $\Delta S_{\text{info}} =
   -k_B\ln 2$ on erasure forces $\Delta S_{\text{phys}} \geq k_B \ln 2$, hence
   $\Delta Q \geq k_B T \ln 2$.
3. **M/E/I chains two equivalences**: $E = mc^2$ and $E = k_B T \ln 2$ per bit, giving
   $m_{\text{bit}} = k_B T \ln 2 / c^2$.
4. **The mass is minute** — $3.19\times10^{-38}$ kg, tens of millions of times lighter than an
   electron. A terabyte of data would carry around $10^{-25}$ kg.
5. **The $T \to 0$ limit is clean**: zero temperature gives zero mass, read as information being
   unable to exist at absolute zero — which the third law puts out of reach regardless.
6. **The steps differ in standing.** Landauer is established and measured; M/E/I is a published
   proposal resting on an additional assumption; the fifth-state claim is speculation built on that.
